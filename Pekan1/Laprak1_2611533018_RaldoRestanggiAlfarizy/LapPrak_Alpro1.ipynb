## Tugas Praktikum

**Soal : Membuat sebuah file program Python bebas dengan nama tugas1_nim.py.**


```python
nama = "Raldo Restanggi Alfarizy"
nim = "2611533018"
jurusan = "Informatika"
universitas = "Universitas Andalas"
angkatan = "2026"
domisili = "Padang"

print("=== BIODATA DIRI ===")
print("Nama        :", nama)
print("NIM         :", nim)
print("Jurusan     :", jurusan)
print("Universitas :", universitas)
print("Angkatan    :", angkatan)
print("Domisili    :", domisili)
```

**Penjelasan :** Program ini dibuat menggunakan bahasa pemrograman Python untuk menampilkan data diri penulis. Pada bagian awal, beberapa variabel dideklarasikan untuk menyimpan data diri seperti nama, NIM, jurusan, universitas, angkatan, dan domisili, di mana masing-masing nilai berupa tipe data string sehingga diapit oleh tanda kutip. Setelah variabel-variabel tersebut dideklarasikan, fungsi print() digunakan untuk menampilkan data tersebut ke layar, di mana setiap baris print() memuat teks label yang dipisahkan dengan tanda koma dari variabel yang bersangkutan, sehingga saat program dijalankan, keluaran yang muncul berupa judul "DATA DIRI" diikuti oleh nama, NIM, jurusan, universitas, angkatan, dan domisili secara berurutan sesuai dengan urutan pemanggilan fungsi print() pada program.
