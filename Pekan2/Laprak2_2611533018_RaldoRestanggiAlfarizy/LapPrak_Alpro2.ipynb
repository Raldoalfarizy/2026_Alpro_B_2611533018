## 2.1 Landasan Teori

**1. Pengenalan Bahasa Pemrograman Python**

Python merupakan salah satu bahasa pemrograman tingkat tinggi yang menggunakan konsep interpreted, object-oriented, dan dynamically typed. Python pertama kali dikembangkan oleh Guido van Rossum pada tahun 1991. Bahasa ini dirancang dengan sintaks yang sederhana dan mudah dibaca, sehingga cukup banyak digunakan untuk pembelajaran maupun pengembangan program. Python dapat diterapkan dalam berbagai bidang, seperti pembuatan perangkat lunak, otomatisasi, analisis data, dan kecerdasan buatan.

​**2. Konsep Variabel dalam Python**

Variabel digunakan untuk menyimpan sebuah nilai yang nantinya dapat digunakan kembali dalam program. Pada Python, variabel dapat dibuat secara langsung dengan memberikan nilai menggunakan tanda = tanpa harus menentukan tipe datanya terlebih dahulu. Dalam penulisannya, nama variabel memiliki beberapa aturan, seperti membedakan huruf besar dan kecil (case-sensitive), tidak boleh diawali angka, serta tidak dapat menggunakan kata kunci yang sudah digunakan oleh Python.

**​3. Tipe Data Dasar**

Tipe data menunjukkan jenis nilai yang disimpan dalam sebuah variabel. Python memiliki beberapa tipe data dasar yang sering digunakan, yaitu:

- Integer (int): digunakan untuk menyimpan bilangan bulat, seperti -5, 0, dan 100.

- Float (float): digunakan untuk menyimpan bilangan yang memiliki nilai desimal, seperti 3.14 dan -0.01.

- String (str): digunakan untuk menyimpan teks atau kumpulan karakter yang ditulis menggunakan tanda petik, seperti "Praktikum Python".

- Boolean (bool): digunakan untuk menyimpan nilai logika yang terdiri dari True dan False.

**​4. Dynamic Typing dan Type Casting**

Python menggunakan sistem dynamic typing, yaitu tipe data variabel dapat ditentukan secara otomatis berdasarkan nilai yang diberikan ketika program dijalankan. Python juga menyediakan type casting yang digunakan untuk mengubah suatu nilai dari satu tipe data ke tipe lainnya. Contohnya adalah penggunaan int(), float(), dan str(). Pemahaman mengenai type casting diperlukan agar data dapat digunakan sesuai dengan operasi yang dilakukan dan mengurangi kemungkinan terjadinya type error.

## 2.2 Kode Program

### 2.2.1 Program "Boolean_2611533018"

```python
# Buat file dengan nama Boolean_NIM.py
# Nama variabel ditambah 4 digit terakhir NIM contoh : a_3018
# Deklarasi variabel dengan tipe data boolean
is_lulus_3018 = True
is_cumlaude_3018 = True

# Mengunakan Boolean
nilai_3018 = 85
batas_lulus_3018 = 75

# Menentukan nilai Boolean dari kondisi
status_kelulusan_3018 = nilai_3018 >= batas_lulus_3018 # Hasilnya akan True

print("=== Check kelulusan ===")
print("Nilai : ", nilai_3018)
print("Apakah lulus? : ", status_kelulusan_3018)
if is_lulus_3018 and is_cumlaude_3018:
    print("Selamat, Anda lulus dengan predikat Cumlaude!")
```

**Penjelasan :** Program ini mendemonstrasikan tipe data boolean, dimulai dari dua variabel boolean langsung (True), lalu membandingkan nilai numerik memakai operator perbandingan (>=) yang otomatis menghasilkan nilai boolean pada variabel status_kelulusan, dan diakhiri pernyataan if dengan operator logika and untuk mengecek dua kondisi sekaligus sebelum menampilkan pesan kelulusan.

3. Bagian 5 berfungsi untuk menentukan status_kelulusan_3018 
4. Bagian 6 sampai 8 berfungsi untuk menampilkan judul, nilai mahasiswa, dan hasil status kelulusan ke layar yang menggunakan operator logika if.

### 2.2.2 Program "Input_2611533018"

```python
# Buat file dengan nama Input_NIM.py
# Buat program untuk menghitung luas persegi panjang
# Nama variabel ditambah 4 digit terakhir NIM contoh : a_3018
# Program ini menggunakan fungsi input() untuk meminta pengguna
# Masukkan nilai panjang dan lebar persegi panjang
# Nilai yang dimasukkan akan dikonversi mejadi data integer

panjang_3018 = input("Masukkan nilai panjang: ")
lebar_3018 = input("Masukkan nilai lebar: ")
luas_3018 = int(panjang_3018) * int(lebar_3018)
print("Luas = ", luas_3018)
```

**Penjelasan :** Program ini menghitung luas persegi panjang dengan fungsi input() untuk menerima panjang dan lebar dari pengguna, di mana nilainya masih berupa string sehingga dikonversi ke integer dengan int() sebelum dikalikan, lalu hasilnya ditampilkan dengan print().

### 2.2.3 Program "Konstanta_2611533018"

```python
# Buat file dengan nama Konstanta_NIM.py
# Program ini menggunakan konstanta untuk menghitung luas lingkaran
# Nama variabel ditambah 4 digit terakhir NIM contoh : jari_3018

from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_3018 = float(input("Masukkan nilai jari-jari: "))
luas_3018 = PI * jari_3018 * jari_3018
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3018, luas_3018))
```

**Penjelasan :** Program ini menghitung luas lingkaran menggunakan konstanta PI yang dideklarasikan dengan anotasi Final dari modul typing agar nilainya tidak bisa diubah, kemudian jari-jari diminta lewat input() dan dikonversi ke float, lalu hasilnya ditampilkan dengan format string (%f dan %.2f) agar angka desimalnya rapi.

1. Bagian 1 sampai 2 berfungsi untuk mengimpor Final dan menentukan nilai konstanta PI sebesar 3.14.
2. Bagian 3 sampai 4 berfungsi untuk menampilkan nilai PI dan meminta pengguna memasukkan nilai jari-jari lingkaran.
3. Bagian 5 berfungsi untuk menghitung luas lingkaran menggunakan nilai PI dan jari-jari yang telah dimasukkan.
4. Bagian 6 berfungsi untuk menampilkan hasil perhitungan luas lingkaran dengan format dua angka di belakang koma.

### 2.2.4 Program "Numerik_2611533018"

```python
# buat file dengan nama Numerik_NIm.py
# Nama variabel ditambah 4 digit terakhir NIM contoh : a_3018
# create a variable with numeric type of value.
a_3018 = 100
print("The type of variable having value", a_3018, "is", type(a_3018))

#create a variable with float value
b_3018 = 10.2345
print("The type of variable having value", b_3018, "is", type(b_3018))

#create a variable with complex value.
c_3018 = 100+3j
print("The type of variable having value", c_3018, "is", type(c_3018))
```

**Penjelasan :** Program ini menunjukkan tiga jenis tipe data numerik Python — integer, float, dan complex — di mana tiap nilai disimpan dalam variabel lalu ditampilkan bersama tipe datanya menggunakan fungsi type() untuk membuktikan Python otomatis mengenali tipe data yang dipakai.
