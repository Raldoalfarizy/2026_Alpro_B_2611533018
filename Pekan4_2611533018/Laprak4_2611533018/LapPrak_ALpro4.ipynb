### 2.1 Dasar Teori

**a. If Tunggal**
Pernyataan `if` tunggal digunakan ketika suatu blok kode hanya perlu dijalankan jika satu kondisi bernilai *True*. Jika kondisi bernilai *False*, blok kode di dalam `if` akan dilewati begitu saja tanpa ada alternatif lain yang dijalankan.

**b. If-Elif-Else**
Struktur `if-elif-else` digunakan ketika terdapat beberapa kondisi yang saling eksklusif (hanya satu kondisi yang boleh terpenuhi). Python akan mengevaluasi kondisi secara berurutan dari atas, dan begitu satu kondisi bernilai *True*, blok tersebut dijalankan lalu seluruh percabangan lain otomatis diabaikan. Operator logika seperti `and`, `or`, dan `!=` (tidak sama dengan) sering digunakan untuk menggabungkan lebih dari satu syarat dalam satu kondisi.

**c. Multi-If Terpisah**
Berbeda dengan `if-elif-else`, multi-if terpisah terdiri atas beberapa pernyataan `if` yang berdiri sendiri-sendiri. Setiap `if` dievaluasi secara independen tanpa memandang hasil evaluasi `if` sebelumnya. Bentuk ini cocok digunakan ketika beberapa kondisi *boleh* terpenuhi sekaligus dan efeknya perlu diakumulasikan, misalnya pada kasus penjumlahan diskon bertingkat.

**d. Nested If**
Nested if adalah percabangan yang diletakkan di dalam blok percabangan lain. Struktur ini digunakan untuk melakukan pengecekan bertingkat, misalnya mengelompokkan suatu nilai ke dalam beberapa rentang (range) sekaligus, seperti pada konversi nilai angka menjadi nilai huruf.

**e. Match-Case**
Pernyataan `match-case` merupakan fitur pattern matching yang tersedia sejak Python 3.10. Struktur ini membandingkan satu nilai (subjek) dengan beberapa pola (`case`) yang telah didefinisikan, dan akan mengeksekusi blok `case` pertama yang cocok. Pola `case _` digunakan sebagai kondisi default apabila tidak ada pola lain yang cocok, serupa dengan `else` pada `if-elif-else`. Struktur ini umumnya lebih ringkas dan mudah dibaca dibandingkan `if-elif-else` yang panjang ketika membandingkan satu variabel dengan banyak nilai diskrit.

#### 2.2.1 `if1_2611533018.py` — If Tunggal (Kelulusan IPK)

Program ini merupakan latihan paling sederhana yang hanya menggunakan satu blok `if` tanpa `else`. Program membaca IPK yang diinput pengguna, lalu hanya mencetak pesan kelulusan apabila IPK yang dimasukkan lebih besar dari 2.75. Jika syarat tidak terpenuhi, program tidak mencetak apa-apa selain baris "Program selesai" karena tidak ada blok alternatif yang disediakan.

```python
# Buat file dengan nama if1_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Pogram ini menggunakan fungsi input ()


ipk_3018 = float(input("Input IPK Anda = "))

if ipk_3018 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_3018))

print("Program selesai")
```

**Penjelasan:** Karena nilai `ipk_3018` yang diinput (3.5) lebih besar dari 2.75, kondisi `if` bernilai *True* sehingga pesan kelulusan dicetak. Apabila IPK yang dimasukkan ≤ 2.75, baris pesan tersebut akan dilewati dan program langsung mencetak "Program selesai".

#### 2.2.2 `if2_2611533018.py` — If-Else (Kelulusan IPK)

Program ini merupakan pengembangan dari `if1_2611533018.py` dengan menambahkan blok `else`, sehingga terdapat dua kemungkinan keluaran: pesan kelulusan jika IPK > 2.75, atau pesan "Anda Tidak Lulus" jika sebaliknya. Dengan `else`, program menjamin salah satu dari dua blok pasti dijalankan.

```python
# Buat file dengan nama if2_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Pogram ini menggunakan fungsi input ()


ipk_3018 = float(input("Input IPK Anda = "))

if ipk_3018 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_3018))
else:
    print("Anda Tidak Lulus")
print("Program selesai")
```

**Penjelasan:** IPK yang dimasukkan (2.5) tidak lebih besar dari 2.75, sehingga kondisi `if` bernilai *False* dan eksekusi program berpindah ke blok `else` yang mencetak "Anda Tidak Lulus". Dibandingkan dengan `if1_2611533018.py`, versi ini lebih baik karena selalu memberikan keluaran yang jelas untuk kedua kemungkinan IPK.

#### 2.2.3 `if_elif_else1_2611533018.py` — If-Elif-Else dengan Operator Logika (Umur & SIM)

Program ini menggabungkan dua input (umur dan kepemilikan SIM) menjadi empat kemungkinan kondisi menggunakan `if-elif-elif-else` beserta operator logika `and` dan `!=`. Karena menggunakan `elif`, keempat kondisi bersifat saling eksklusif — hanya satu blok yang akan dijalankan untuk setiap kombinasi input.

```python
# Buat file dengan nama if_elif_else1_2611533018.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Pogram ini menggunakan fungsi input ()

umur_3018 = int(input("Input Umur Anda = "))
sim_3018 = input("Apakah Anda Memiliki SIM (y/t) = ") [0]

if umur_3018 >= 17 and sim_3018 == 'y':
    print("Anda sudah dewasa dan boleh bawa motor")
elif umur_3018 >= 17 and sim_3018 != 'y':
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")
elif umur_3018 < 17 and sim_3018 == 'y':
    print("Anda belum cukup umur punya SIM")
else: 
    print("Anda belum cukup umur dan tidak boleh bawa motor")
print("Program selesai")
```

**Penjelasan:** Dengan umur 20 (≥ 17) dan status SIM `'y'`, kondisi pertama `umur_3018 >= 17 and sim_3018 == 'y'` langsung bernilai *True*, sehingga pesan "Anda sudah dewasa dan boleh bawa motor" dicetak dan ketiga kondisi `elif`/`else` lainnya tidak dievaluasi lagi.

#### 2.2.4 `multi_if1_2611533018.py` — Multi-If Terpisah (Umur & SIM)

Program ini membahas kasus yang sama seperti pada `if_elif_else1_2611533018.py`, namun keempat kondisi ditulis sebagai empat pernyataan `if` yang berdiri sendiri-sendiri (bukan `elif`). Karena keempat kondisi yang dirancang bersifat saling meniadakan (mutually exclusive) berdasarkan kombinasi umur dan SIM, hasil akhirnya tetap sama dengan versi `if-elif-else`, meskipun secara teknis keempat `if` tetap dievaluasi satu per satu tanpa saling melompati.

```python
# Buat file dengan nama multi_if1_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Pogram ini menggunakan fungsi input ()

umur_3018 = int(input("Input Umur Anda = "))
sim_3018 = input("Apakah Anda Memiliki SIM (y/t) = ") [0]

if umur_3018 >= 17 and sim_3018 == 'y':
    print("Anda sudah dewasa dan boleh bawa motor")

if umur_3018 >= 17 and sim_3018 != 'y':
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")

if umur_3018 < 17 and sim_3018 == 'y':
    print("Anda belum cukup umur punya SIM")

if umur_3018 < 17 and sim_3018 != 'y':
    print("Anda belum cukup umur bawa motor")
```

**Penjelasan:** Dengan umur 16 (< 17) dan status SIM `'t'`, hanya kondisi keempat `umur_3018 < 17 and sim_3018 != 'y'` yang bernilai *True*, sehingga hanya pesan "Anda belum cukup umur bawa motor" yang dicetak. Perlu diperhatikan bahwa karena setiap `if` dievaluasi secara terpisah (bukan `elif`), penggunaan multi-if pada kasus yang bersifat eksklusif seperti ini sebenarnya kurang efisien dibandingkan `if-elif-else`, karena Python tetap memeriksa keempat kondisi meskipun kondisi yang benar sudah ditemukan lebih awal.

#### 2.2.5 `multi_if2_2611533018.py` — Multi-If Terpisah (Diskon Belanja Akumulatif)

Berbeda dengan kasus sebelumnya, program ini menunjukkan penggunaan multi-if terpisah yang memang tepat digunakan, yaitu untuk mengakumulasikan beberapa diskon sekaligus. Terdapat tiga pernyataan `if` independen: diskon belanja besar (> Rp100.000), diskon member, dan diskon kode promo. Karena masing-masing `if` dievaluasi sendiri-sendiri, persentase diskon dari ketiganya dapat dijumlahkan (ditumpuk) apabila lebih dari satu syarat terpenuhi bersamaan.

```python
# Buat file dengan nama multi_if2_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Pogram ini menggunakan fungsi input ()
# Program menghitung diskon belanja

# input dari user
total_belanja_3018 = int(input("Masukkan total belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_3018 = input("Apakah Anda member (y/t): ").strip().lower()
is_member_3018 = input_member_3018 in ['y', 'ya']

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_3018 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_3018 = input_promo_3018 in ['y', 'ya']

total_diskon_persen_3018 = 0

# Multi-IF terpisah: setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_3018 > 100000:
    total_diskon_persen_3018 += 10  # Diskon belanja besar

if is_member_3018:
    total_diskon_persen_3018 += 5  # Diskon member

if kode_promo_valid_3018:
    total_diskon_persen_3018 += 15  # Diskon Voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_3018 = total_belanja_3018 * total_diskon_persen_3018 / 100
total_bayar_3018 = total_belanja_3018 - nominal_diskon_3018

# Output hasil
print("---Rincian Pembayaran---")
print(f"Total Diskon :  {total_diskon_persen_3018}% (Rp {nominal_diskon_3018:,.0f})") 
print(f"Total bayar : Rp {total_bayar_3018:,.0f}")

print(f"Total diskon yang anda dapatkan: {total_diskon_persen_3018}%")
# Output: Total diskon yang anda dapatkan: 30% jika belanja > 100000, member, dan kode promo valid
```

**Penjelasan:** Total belanja Rp150.000 memenuhi syarat diskon belanja besar (+10%), status member `'y'` memenuhi diskon member (+5%), dan kode promo valid `'y'` memenuhi diskon voucher (+15%), sehingga ketiga diskon terakumulasi menjadi 30%. Nominal diskon dihitung sebesar 150.000 × 30% = Rp45.000, sehingga total yang harus dibayar adalah Rp105.000. Kasus ini menunjukkan bahwa multi-if terpisah adalah pilihan yang tepat ketika beberapa kondisi memang boleh terpenuhi dan efeknya perlu digabungkan, berbeda dengan kasus pada Subbab 2.2.4.

#### 2.2.6 `nilai_2611533018.py` — Nested If / Percabangan Berjenjang (Konversi Nilai ke Huruf)

Program ini mengonversi nilai angka menjadi nilai huruf (A, B, C, D, E) menggunakan rangkaian `if-elif-else` yang dievaluasi secara berjenjang dari batas tertinggi ke batas terendah. Setiap `elif` hanya akan diperiksa apabila seluruh kondisi di atasnya bernilai *False*, sehingga logika ini secara konsep setara dengan percabangan bersarang (nested if) untuk pengelompokan nilai ke dalam beberapa rentang.

```python
# Buat file dengan nama nilai_nim.py
# Buat program untuk nested if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Pogram ini menggunakan fungsi input ()
# Program konversi nilai angka menjadi huruf

nilai_3018 = int(input("Masukkan nilai angka: "))

if nilai_3018 >= 81:
    print("A")
elif nilai_3018 >= 70:
    print("B")
elif nilai_3018 >= 60:
    print("C")
elif nilai_3018 >= 50:
    print("D")
else:
    print("E")
```

**Penjelasan:** Nilai 75 diperiksa mulai dari kondisi teratas: `nilai_3018 >= 81` bernilai *False* (karena 75 < 81), sehingga berlanjut ke `elif nilai_3018 >= 70` yang bernilai *True* (karena 75 ≥ 70), maka huruf "B" dicetak dan pemeriksaan kondisi berikutnya (`>= 60`, `>= 50`, dan `else`) tidak lagi dilakukan. Urutan pemeriksaan dari batas tertinggi ke terendah ini penting agar setiap nilai hanya jatuh ke satu kategori yang benar.

#### 2.2.7 `match_case1_2611533018.py` — Match-Case (Konversi Angka Bulan)

Program ini mengonversi angka bulan (1–12) menjadi nama bulan menggunakan pernyataan `match-case`. Setiap `case` mewakili satu kemungkinan nilai angka bulan, sedangkan `case _` berfungsi sebagai kondisi default untuk menangani input yang tidak valid (di luar rentang 1–12).

```python
# Buat file dengan nama match_case1_nim.py
# Buat program untuk match case
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Pogram ini menggunakan fungsi input ()
# Program konversi angka menjadi nama bulan

bulan = int(input("Masukkan angka bulan (1-12): "))

match bulan:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

**Penjelasan:** Nilai `bulan` yang diinput (8) dicocokkan dengan pola `case 8`, sehingga Python langsung mencetak "Agustus" tanpa perlu memeriksa case-case lainnya. Jika pengguna memasukkan angka di luar 1–12, misalnya 13, maka tidak ada pola yang cocok sehingga program akan jatuh ke `case _` dan mencetak "Angka tidak valid". Dibandingkan menuliskan 12 blok `elif` secara manual, struktur `match-case` jauh lebih ringkas dan mudah dibaca untuk kasus dengan banyak nilai diskrit seperti ini.

```python
"""
Tugas 4 - Praktikum Algoritma & Pemrograman
Sistem Loket Terpadu & Audit Transaksi Ekspedisi Wahana
Nama   : Raldo Restanggi Alfarizy
NIM    : 2611533018
Materi : if tunggal, if-elif-else + operator logika, multi-if terpisah
         (diskon akumulasi), dan match-case
"""

print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")


# Input Data Pengunjung & String Handling

nama_pengunjung_3018 = input("Masukkan Nama Pengunjung        : ")
umur_3018 = int(input("Input umur anda                 : "))
sim_3018 = input("Apakah Anda Sudah Punya SIM C (y/t): ").strip().lower()
jumlah_tiket_3018 = int(input("Masukkan jumlah tiket           : "))

# Validasi kelogisan jumlah tiket (if tunggal)
if jumlah_tiket_3018 <= 0:
    print("Peringatan: Kuota tiket tidak valid!")


# Pemilihan Wahana Menggunakan match-case (kasus 1-5 + default _)
print("\nPilihan Paket Wahana (1-5):")
print("  1. Safari Rimba         (Rp 50,000)")
print("  2. Arung Jeram          (Rp 75,000)")
print("  3. Motor ATV Ekstrim    (Rp 120,000)")
print("  4. Roller Coaster Kilat (Rp 100,000)")
print("  5. All-Access VIP       (Rp 220,000)")
paket_3018 = int(input("Masukkan nomor paket (1-5)      : "))

match paket_3018:
    case 1:
        nama_wahana_3018 = "Wahana Safari Rimba"
        harga_satuan_3018 = 50000
    case 2:
        nama_wahana_3018 = "Wahana Arung Jeram"
        harga_satuan_3018 = 75000
    case 3:
        nama_wahana_3018 = "Wahana Motor ATV Ekstrim"
        harga_satuan_3018 = 120000
    case 4:
        nama_wahana_3018 = "Wahana Roller Coaster Kilat"
        harga_satuan_3018 = 100000
    case 5:
        nama_wahana_3018 = "Wahana All-Access VIP"
        harga_satuan_3018 = 220000
    case _:
        print("Paket wahana tidak valid!")
        exit()

is_member_3018 = input("Apakah Anda member? (y/t)       : ").strip().lower()
kode_promo_valid_3018 = input("Apakah kode promo valid? (y/t)  : ").strip().lower()


# Validasi Izin Kendali Wahana
#    - Paket 3 (ATV): if-elif-else dengan operator logika (and, !=)
#    - Paket lain   : if-else sederhana (umur >= 10)
#    Ditulis sebagai SATU rangkaian if-elif-else (tanpa nested-if)
#    dengan menggabungkan syarat paket_3018 == 3 di setiap kondisi.

print("\n--- KELAYAKAN PENGENDARA WAHANA ---")

if paket_3018 == 3 and umur_3018 >= 17 and sim_3018 == 'y':
    print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
elif paket_3018 == 3 and umur_3018 >= 17 and sim_3018 != 'y':
    print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
elif paket_3018 == 3 and umur_3018 < 17 and sim_3018 == 'y':
    print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")
elif paket_3018 == 3:
    print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")
elif umur_3018 >= 10:
    print(f"Status Akses: Anda memenuhi syarat umur untuk {nama_wahana_3018}.")
else:
    print(f"Status Akses: Anda belum cukup umur untuk {nama_wahana_3018}.")


# 4. Akumulasi Diskon Bertingkat Menggunakan Multi-IF Terpisah
subtotal_3018 = harga_satuan_3018 * jumlah_tiket_3018
total_diskon_persen_3018 = 0

if subtotal_3018 >= 200000:
    total_diskon_persen_3018 += 10          # Diskon Belanja Besar

if is_member_3018 in ['y', 'ya']:
    total_diskon_persen_3018 += 5           # Diskon Member

if kode_promo_valid_3018 in ['y', 'ya']:
    total_diskon_persen_3018 += 15          # Diskon Voucher Promo

if jumlah_tiket_3018 >= 5:
    total_diskon_persen_3018 += 5           # Diskon Tambahan Rombongan

nominal_diskon_3018 = subtotal_3018 * (total_diskon_persen_3018 / 100)
total_bayar_3018 = subtotal_3018 - nominal_diskon_3018


# Evaluasi Kelulusan Audit Menggunakan if-else
if total_bayar_3018 > 300000:
    catatan_layanan_3018 = "Selamat! Anda berhak mendapatkan Souvenir Gratis."
else:
    catatan_layanan_3018 = "Terima kasih telah berkunjung."

print("\n--- Rincian Pembayaran ---")
print(f"Subtotal Belanja : Rp {subtotal_3018:,.0f}")
print(f"Total Diskon     : {total_diskon_persen_3018}% (Rp {nominal_diskon_3018:,.0f})")
print(f"Total Bayar      : Rp {total_bayar_3018:,.0f}")
print(f"Catatan Layanan  : {catatan_layanan_3018}")
print("Program Selesai")
```

#### 2.3.4 Penjelasan

**Validasi jumlah tiket (if tunggal).** Sebelum data lain diproses, program memeriksa `jumlah_tiket_3018 <= 0`. Karena pada kedua skenario uji jumlah tiket yang diinput bernilai positif (5 dan 1), blok peringatan ini tidak tereksekusi. Jika pengguna memasukkan 0 atau bilangan negatif, program tetap akan melanjutkan eksekusi (karena tidak ada `else`/`exit()` pada validasi ini) namun telah menampilkan peringatan kepada pengunjung.

**Pemilihan wahana (match-case).** Nilai `paket_3018` yang diinput dicocokkan dengan `case 1` hingga `case 5`. Pada Skenario 1, `paket_3018 = 5` mencocokkan `case 5` sehingga `nama_wahana_3018` diisi "Wahana All-Access VIP" dan `harga_satuan_3018` diisi 220000. Pada Skenario 2, `paket_3018 = 3` mencocokkan `case 3` sehingga dipilih "Wahana Motor ATV Ekstrim" dengan harga 120000. Apabila input di luar 1–5, `case _` akan tereksekusi dan program dihentikan dengan `exit()`.

**Validasi kelayakan pengendara (if-elif-else gabungan).** Rangkaian kondisi ini menggabungkan syarat khusus untuk paket ATV (`paket_3018 == 3`) dengan syarat umum wahana lain, seluruhnya ditulis sebagai satu rangkaian `if-elif-else` tanpa nested-if, dengan cara menyertakan `paket_3018 == 3` pada setiap cabang ATV. Pada Skenario 1 (bukan ATV, umur 16 ≥ 10), program jatuh ke cabang `elif umur_3018 >= 10` sehingga mencetak status memenuhi syarat umur. Pada Skenario 2 (ATV, umur 16 < 17, SIM `'y'`), program mencocokkan cabang ketiga `paket_3018 == 3 and umur_3018 < 17 and sim_3018 == 'y'`, yang secara khusus menangani kasus "identitas tidak valid" — yaitu pengunjung mengaku memiliki SIM C padahal usianya belum memenuhi syarat hukum untuk memilikinya.

**Akumulasi diskon (multi-if terpisah).** Sama seperti pada `multi_if2_2611533018.py`, keempat pernyataan `if` (belanja besar, member, promo, rombongan) dievaluasi secara independen sehingga diskon dapat ditumpuk. Pada Skenario 1: subtotal Rp1.100.000 (≥ Rp200.000, +10%), member (+5%), promo (+15%), dan jumlah tiket 5 (≥ 5, +5%) — total diskon 35%, sehingga total bayar menjadi Rp715.000. Pada Skenario 2: subtotal Rp120.000 tidak memenuhi satu pun syarat diskon (bukan member, bukan promo, tiket hanya 1), sehingga total diskon tetap 0% dan total bayar sama dengan subtotal.

**Evaluasi souvenir (if-else).** Total bayar dibandingkan dengan ambang batas Rp300.000. Pada Skenario 1, total bayar Rp715.000 > Rp300.000 sehingga pengunjung mendapat pesan souvenir gratis. Pada Skenario 2, total bayar Rp120.000 tidak melebihi ambang batas sehingga hanya dicetak ucapan terima kasih biasa.
