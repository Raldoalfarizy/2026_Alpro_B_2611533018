## 2.1 Dasar Teori

### 2.1.1 Perulangan `for` pada Python

Perulangan `for` pada Python digunakan untuk menelusuri setiap elemen dari sebuah objek *iterable* (misalnya `range`, list, atau string) secara berurutan. Pada setiap putaran, nilai elemen disimpan ke variabel perulangan, lalu blok perintah di dalam `for` dieksekusi.

```python
for variabel in iterable:
    # blok perintah yang diulang (wajib menjorok/indentasi)
```

### 2.1.2 Fungsi `range()`

`range()` menghasilkan deret bilangan bulat yang umum dipakai sebagai *iterable* pada `for`. Nilai `stop` bersifat **eksklusif** (tidak ikut dihasilkan).

| Bentuk | Deret yang dihasilkan | Contoh | Hasil |
|---|---|---|---|
| `range(stop)` | `0` sampai `stop - 1` | `range(5)` | 0, 1, 2, 3, 4 |
| `range(start, stop)` | `start` sampai `stop - 1` | `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(start, stop, step)` | dari `start`, bertambah `step`, berhenti sebelum `stop` | `range(4, 0, -1)` | 4, 3, 2, 1 |

Banyaknya iterasi pada `range(start, stop)` dengan `step = 1` adalah `stop - start`. Jika `stop <= start`, perulangan tidak berjalan sama sekali, misalnya `range(0)`. Hal ini dimanfaatkan pada pola yang jumlah spasinya bisa bernilai nol.

### 2.1.3 Parameter `end` pada `print()`

Secara bawaan `print()` mengakhiri keluaran dengan karakter baris baru (`\n`). Parameter `end` mengubah perilaku tersebut:

* `print("*", end="")` mencetak tanpa pindah baris dan tanpa spasi tambahan.
* `print("*", end=" ")` mencetak lalu menambah satu spasi.
* `print()` tanpa argumen hanya berfungsi memindahkan kursor ke baris berikutnya, dan dipakai di akhir setiap baris pola.

### 2.1.4 Nested for (Perulangan Bersarang)

*Nested for* adalah perulangan `for` yang berada di dalam badan perulangan `for` lain. Untuk setiap satu putaran perulangan luar, perulangan dalam dijalankan **sampai selesai**.

```python
for baris in range(1, 4):          # perulangan luar  -> mengatur baris
    for kolom in range(1, 5):      # perulangan dalam -> mengatur kolom
        print("*", end=" ")
    print()                        # pindah baris setelah kolom selesai
```

Jika perulangan luar berjalan sebanyak *m* kali dan perulangan dalam sebanyak *n* kali, badan perulangan dalam dieksekusi sebanyak **m × n** kali. Bila jumlah elemen pada tiap baris bergantung pada nomor baris (misalnya segitiga), batas `range` perulangan dalam dinyatakan sebagai fungsi dari variabel perulangan luar, yaitu *f(baris)*.

### 2.1.5 Percabangan di dalam Perulangan dan Operator Modulus

Pernyataan `if` dapat ditempatkan di dalam perulangan untuk menyaring atau membedakan perlakuan tiap iterasi. Operator modulus `%` menghasilkan sisa bagi, sehingga `i % 2 == 0` bernilai benar bila `i` genap. Kondisi seperti `i == 1 or i == tinggi` dapat dipakai untuk mengenali baris pertama dan terakhir pada pola berbingkai.

### 2.1.6 Variabel Akumulator

Variabel akumulator menampung hasil yang terus bertambah selama perulangan, misalnya `jumlah = jumlah + i` dengan nilai awal `jumlah = 0`. Untuk deret bilangan asli `1 + 2 + ... + n` berlaku rumus **n(n+1)/2**, dan untuk jumlah bilangan genap dari 1 sampai *n* dengan `m = n // 2` berlaku **m(m+1)**. Rumus ini dipakai untuk memverifikasi hasil program.

## 2.2 Kode Program: Pola Jam Pasir Kristal Palindromik Berbingkai (`tugas5_2611533018.py`)

### Deskripsi Soal

Dibuat program konsol yang menerima bilangan bulat positif **N** (skala angka tertinggi pada puncak dan alas jam pasir), lalu mencetak pola jam pasir berbingkai yang terdiri dari **3 fase dan 2 bingkai horizontal**. Ketentuan teknisnya:

* seluruh variabel berakhiran 4 digit terakhir NIM;
* masukan diambil dengan `input()` dan dikonversi dengan `int()`;
* seluruh bingkai, deret spasi, dan deret angka dibentuk dengan perulangan `for` bersarang berparameter `end=""`, **tanpa** perkalian string seperti `"=" * n`;
* setiap baris pola diakhiri `print()` tanpa argumen.

### Kode Program

````python
# Nama   : Raldo Restanggi Alfarizy
# NIM    : 2611533018
# Tugas  : Praktikum Alpro Pekan 5 - Pola Jam Pasir Kristal Palindromik Berbingkai
# Catatan: seluruh variabel berakhiran 4 digit terakhir NIM (3018)
#          dan semua pola dibentuk dengan for bersarang (tanpa perkalian string).

print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")
n_3018 = int(input("Masukkan ukuran skala jam pasir (N): "))

# Lebar isi bingkai = 4 * N + 5 karakter '='
lebar_3018 = 4 * n_3018 + 5

# Bingkai Atas
print("#", end="")
for garis_3018 in range(lebar_3018):
    print("=", end="")
print("#")

# Fase 1: Jam Pasir Atas (N turun s.d. 1) 
for baris_3018 in range(n_3018, 0, -1):
    print("| ", end="")                                   # pagar kiri + 1 spasi padding

    for spasi_kiri_3018 in range(2 * (n_3018 - baris_3018)):  # spasi penyeimbang kiri
        print(" ", end="")

    for angka_mundur_3018 in range(baris_3018, 0, -1):    # deret mundur: baris ... 1
        print(angka_mundur_3018, end=" ")

    print("<*>", end="")                                   # poros kristal

    for angka_maju_3018 in range(1, baris_3018 + 1):      # deret maju: 1 ... baris
        print(" ", end="")
        print(angka_maju_3018, end="")

    for spasi_kanan_3018 in range(2 * (n_3018 - baris_3018)):  # spasi penyeimbang kanan
        print(" ", end="")

    print(" |", end="")                                   # 1 spasi padding + pagar kanan
    print()

# Fase 2: Poros Titik Pusat 
print("|", end="")
for spasi_kiri_3018 in range(2 * n_3018 + 1):
    print(" ", end="")
print("<*>", end="")
for spasi_kanan_3018 in range(2 * n_3018 + 1):
    print(" ", end="")
print("|")

# Fase 3: Jam Pasir Bawah (1 naik s.d. N) 
for baris_3018 in range(1, n_3018 + 1):
    print("| ", end="")

    for spasi_kiri_3018 in range(2 * (n_3018 - baris_3018)):
        print(" ", end="")

    for angka_mundur_3018 in range(baris_3018, 0, -1):
        print(angka_mundur_3018, end=" ")

    print("<*>", end="")

    for angka_maju_3018 in range(1, baris_3018 + 1):
        print(" ", end="")
        print(angka_maju_3018, end="")

    for spasi_kanan_3018 in range(2 * (n_3018 - baris_3018)):
        print(" ", end="")

    print(" |", end="")
    print()

# Bingkai Bawah
print("#", end="")
for garis_3018 in range(lebar_3018):
    print("=", end="")
print("#")
````

### Penjelasan

#### a. Variabel yang digunakan

| Variabel | Fungsi |
|---|---|
| `n_3018` | ukuran skala jam pasir (nilai N dari pengguna) |
| `lebar_3018` | jumlah karakter `=` pada bingkai, bernilai `4 * n_3018 + 5` |
| `garis_3018` | penghitung perulangan untuk mencetak bingkai |
| `baris_3018` | nomor baris sekaligus angka terbesar pada baris tersebut |
| `spasi_kiri_3018`, `spasi_kanan_3018` | penghitung perulangan spasi penyeimbang kiri dan kanan |
| `angka_mundur_3018` | penghitung deret angka menurun di sisi kiri poros |
| `angka_maju_3018` | penghitung deret angka menaik di sisi kanan poros |

#### b. Analisis formula matematis

Misalkan **N** adalah ukuran jam pasir dan **b** adalah nomor baris (`baris_3018`) pada Fase 1 maupun Fase 3, dengan `1 ≤ b ≤ N`.

**1. Banyak angka pada tiap sisi poros.** Baris ke-*b* memuat deret `b, b-1, ..., 1` di kiri dan `1, 2, ..., b` di kanan, sehingga masing-masing sisi berisi **b angka**. Karena itu perulangan angka memakai `range(baris, 0, -1)` (menurun) dan `range(1, baris + 1)` (menaik).

**2. Spasi penyeimbang = 2(N − b).** Setiap angka beserta spasi pemisahnya menempati **2 karakter**. Baris terlebar (b = N) tidak butuh spasi penyeimbang, sedangkan baris dengan nomor lebih kecil memiliki (N − b) angka lebih sedikit di tiap sisi. Selisih tersebut digantikan spasi sebanyak 2 × (N − b), yang dibuat dengan perulangan `range(2 * (n_3018 - baris_3018))`. Spasi ini membuat poros `<*>` selalu berada tepat di tengah dan bentuk menyempit ke arah pusat.

**3. Lebar isi bingkai = 4N + 5.** Lebar isi satu baris Fase 1 atau Fase 3 (di antara dua garis tegak `|`) dihitung sebagai berikut:

| Komponen | Banyak karakter |
|---|---|
| 1 spasi padding kiri | 1 |
| spasi penyeimbang kiri | 2(N − b) |
| deret angka mundur (b angka + b spasi) | 2b |
| poros kristal `<*>` | 3 |
| deret angka maju (b spasi + b angka) | 2b |
| spasi penyeimbang kanan | 2(N − b) |
| 1 spasi padding kanan | 1 |
| **Total** | **4N + 5** *(suku b saling menghapus)* |

Penjabarannya: 1 + 2(N − b) + 2b + 3 + 2b + 2(N − b) + 1 = 4N − 4b + 4b + 5 = **4N + 5**. Karena suku *b* saling menghapus, lebar isi **selalu sama pada setiap baris**, dan angka inilah yang menjadi banyaknya karakter `=` pada bingkai.

**4. Baris poros pusat (Fase 2).** Isinya terdiri atas (2N + 1) spasi, `<*>` (3 karakter), dan (2N + 1) spasi, sehingga totalnya (2N + 1) + 3 + (2N + 1) = **4N + 5**, yaitu sama dengan lebar isi baris lainnya.

**5. Ukuran keseluruhan pola.**

| Besaran | Rumus | N = 4 | N = 3 |
|---|---|---|---|
| Karakter `=` pada bingkai | 4N + 5 | 21 | 17 |
| Lebar satu baris (termasuk 2 pembatas) | 4N + 7 | 23 | 19 |
| Jumlah baris keseluruhan (2 bingkai + N + 1 + N) | 2N + 3 | 11 | 9 |

Dengan demikian, jumlah karakter yang dicetak adalah (2N + 3)(4N + 7) dan waktu eksekusinya tumbuh sebanding dengan N² (kompleksitas **O(N²)**), karena baris berjumlah ~2N dan karakter per baris ~4N.

#### c. Tabel penelusuran (N = 4, Fase 1)

| baris (b) | spasi kiri 2(N−b) | deret mundur | poros | deret maju | spasi kanan 2(N−b) |
|:-:|:-:|:-:|:-:|:-:|:-:|
| 4 | 0 | 4 3 2 1 | `<*>` | 1 2 3 4 | 0 |
| 3 | 2 | 3 2 1 | `<*>` | 1 2 3 | 2 |
| 2 | 4 | 2 1 | `<*>` | 1 2 | 4 |
| 1 | 6 | 1 | `<*>` | 1 | 6 |

Fase 3 memakai tabel yang sama dengan urutan baris dibalik (b = 1 sampai N).

#### d. Alur logika program

1. Cetak judul program, lalu baca `n_3018` dengan `int(input(...))`.
2. Hitung `lebar_3018 = 4 * n_3018 + 5`.
3. **Bingkai atas:** cetak `#`, ulangi `lebar_3018` kali untuk mencetak `=`, cetak `#`, lalu pindah baris.
4. **Fase 1:** perulangan luar `for baris_3018 in range(n_3018, 0, -1)` (N turun ke 1). Untuk setiap baris, secara berurutan program mencetak:
   `| ` → spasi kiri → angka mundur → `<*>` → angka maju → spasi kanan → ` |` → `print()`.
5. **Fase 2:** cetak `|`, (2N + 1) spasi, `<*>`, (2N + 1) spasi, `|` tanpa perulangan luar karena hanya satu baris.
6. **Fase 3:** perulangan luar `for baris_3018 in range(1, n_3018 + 1)` (1 naik ke N) dengan isi baris yang identik dengan Fase 1.
7. **Bingkai bawah:** sama dengan bingkai atas.

Struktur *nested for* terdapat pada Fase 1 dan Fase 3: perulangan luar mengatur baris, sedangkan empat perulangan dalam yang berurutan (bukan bertumpuk satu sama lain) mencetak spasi kiri, angka mundur, angka maju, dan spasi kanan. Batas perulangan dalam bergantung pada `baris_3018`, sehingga pola menyesuaikan diri secara dinamis terhadap nilai N yang dimasukkan.

### Latihan 1: `perulangan_for1_2611533018.py`

**Soal:** Buatlah program yang meminta jumlah perulangan dari pengguna, lalu mencetak kalimat "Perulangan ke - i" sebanyak jumlah tersebut dengan `i` dimulai dari 0.

**Kode Jawaban:**

````python
# Buat file dengan nama perulangan_for1_2611533018.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input ()

ulang_3018 = int(input("Masukkan jumlah perulangan = "))

for i_3018 in range(ulang_3018):
    print(f"Perulangan ke - {i_3018}")
````

**Penjelasan:**

* `range(ulang_3018)` menghasilkan deret `0` sampai `ulang_3018 - 1`, sehingga terjadi tepat `ulang_3018` iterasi (untuk masukan 5: 0, 1, 2, 3, 4).
* Pada tiap iterasi nilai `i_3018` disisipkan ke kalimat memakai *f-string* `f"Perulangan ke - {i_3018}"`. Karena `range` dimulai dari 0, penomoran juga dimulai dari 0.

### Latihan 2: `perulangan_for2_2611533018.py`

**Soal:** Buatlah program yang mencetak deret bilangan dari 0 sampai (n − 1) dan dari 1 sampai n pada baris terpisah, sehingga terlihat perbedaan penggunaan `range` satu argumen dan dua argumen.

**Kode Jawaban:**

````python
# Buat file dengan nama perulangan_for2_2611533018.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input ()

ulang_3018 = int(input("Masukkan jumlah perulangan = "))
print("Perulangan ke - 0 sampai ke -",ulang_3018-1)
for i_3018 in range (ulang_3018):
    print(i_3018, end = " ")
print()
print("Perulangan ke - 1 sampai ke - ",ulang_3018)
for i_3018 in range (1,ulang_3018+1):
    print(i_3018, end = " ")
````

**Penjelasan:**

* Perulangan pertama memakai `range(ulang_3018)` yang menghasilkan 0 sampai `ulang_3018 - 1`.
* Perulangan kedua memakai `range(1, ulang_3018 + 1)` yang menghasilkan 1 sampai `ulang_3018`. Batas atas ditulis `ulang_3018 + 1` karena nilai `stop` tidak ikut dihasilkan.
* Kedua perulangan sama-sama berjalan `ulang_3018` kali, perbedaannya hanya pada titik awal deret.
* `print(i_3018, end=" ")` menjaga seluruh angka tetap dalam satu baris, dan `print()` di antara kedua perulangan memindahkan kursor ke baris berikutnya.

### Latihan 3: `Perulangan_for3_2611533018.py`

**Soal:** Buatlah program yang menampilkan penjumlahan bilangan 1 sampai n beserta hasil totalnya, misalnya `1 + 2 + 3 + 4 + 5 = 15`.

**Kode Jawaban:**

````python
# Buat file dengan nama perulangan_for3_2611533018.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input ()

ulang_3018 = int(input("Masukkan jumlah perulangan = "))

jumlah_3018 = 0
for i_3018 in range(1, ulang_3018 + 1):
    print(i_3018, end = " ")
    jumlah_3018 = jumlah_3018 + i_3018

    if i_3018 < ulang_3018:
        print("+", end = " ")
    else:
        print("= ", jumlah_3018, end = " ")
print()
print("Jumlah =", jumlah_3018)
````

**Penjelasan:**

* `jumlah_3018` adalah variabel akumulator dengan nilai awal 0, ditambah `i_3018` pada setiap iterasi.
* Pada setiap iterasi, program mencetak `i_3018`. Bila `i_3018 < ulang_3018` program mencetak tanda `+`, dan pada iterasi terakhir (`i_3018 == ulang_3018`) program mencetak tanda `=` diikuti hasil jumlah. Dengan cara ini tanda `+` tidak muncul setelah angka terakhir.

| i | jumlah sebelum | jumlah sesudah |
|:-:|:-:|:-:|
| 1 | 0 | 1 |
| 2 | 1 | 3 |
| 3 | 3 | 6 |
| 4 | 6 | 10 |
| 5 | 10 | 15 |

* Hasil dapat diverifikasi dengan rumus deret aritmetika n(n+1)/2 = 5 × 6 / 2 = **15**.

### Latihan 4: `jumlah_genap_2611533018.py`

**Soal:** Buatlah program yang menjumlahkan seluruh bilangan genap dari 1 sampai batas yang dimasukkan pengguna, lengkap dengan tampilan penjumlahannya.

**Kode Jawaban:**

````python
# Buat file dengan nama jumlah_genap_2611533018.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input ()

ulang_3018 = int(input("Masukkan nilai batas: "))

jumlah_3018 = 0
for i_3018 in range(1,ulang_3018+1):
    if i_3018 % 2 == 0:
        print(i_3018, end = " ")
        jumlah_3018 = jumlah_3018 + i_3018

        if i_3018 < ulang_3018:
            print("+", end = " ")
        else:
            print(" = ", jumlah_3018, end = " ")
print()
print("Jumlah =", jumlah_3018)
````

**Uji Coba 2 (batas ganjil, batas = 9)**

```text
Masukkan nilai batas: 9
2 + 4 + 6 + 8 + 
Jumlah = 20
```

* Perulangan menelusuri `i_3018` dari 1 sampai `ulang_3018`. Kondisi `if i_3018 % 2 == 0` meneruskan hanya bilangan genap ke blok di dalamnya, sedangkan bilangan ganjil dilewati.
* Untuk bilangan genap, `i_3018` dicetak dan ditambahkan ke `jumlah_3018`. `if` bersarang di dalamnya menentukan apakah dicetak `+` (bila `i_3018 < ulang_3018`) atau `=` beserta hasilnya (bila `i_3018 == ulang_3018`).
* Untuk batas 10: 2 + 4 + 6 + 8 + 10 = 30. Dengan `m = 10 // 2 = 5`, rumus m(m+1) = 5 × 6 = **30** sesuai.
* **Catatan untuk batas ganjil:** pada uji coba 2, nilai `i_3018` terakhir yang genap adalah 8, yang masih lebih kecil dari batas 9. Akibatnya cabang `else` (pencetak tanda `=`) tidak pernah dijalankan, dan tampilan berhenti pada tanda `+`. Nilai `Jumlah` pada baris berikutnya tetap benar: `m = 9 // 2 = 4`, sehingga m(m+1) = 4 × 5 = **20**.

# C.2 Nested for

### Latihan 5: `nested_for1_2611533018.py`

**Soal:** Buatlah program yang mencetak pola angka dengan titik di sisi kiri, di mana jumlah titik berkurang satu pada setiap baris berikutnya.

**Kode Jawaban:**

````python
# Buat file dengan nama nested_for1_2611533018.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input ()

batas_3018 = int(input("masukkan nilai batas: "))
for line_3018 in range(1, batas_3018 + 1):
    for j_3018 in range(1, (-1 * line_3018 + batas_3018) + 1):
        print(".", end = " ")
    print(line_3018)
````

**Penjelasan:**

* Perulangan luar `line_3018` berjalan dari 1 sampai `batas_3018` dan mewakili nomor baris.
* Perulangan dalam `range(1, (-1 * line_3018 + batas_3018) + 1)` menjalankan `batas_3018 - line_3018` iterasi, yaitu banyak titik pada baris tersebut. Setelah titik selesai dicetak, `print(line_3018)` mencetak nomor baris dan pindah baris.
* Rumus jumlah titik pada baris ke-*i* adalah **batas − i**.

| line | jumlah titik (5 − line) |
|:-:|:-:|
| 1 | 4 |
| 2 | 3 |
| 3 | 2 |
| 4 | 1 |
| 5 | 0 |

* Total titik = 4 + 3 + 2 + 1 + 0 = 10, yaitu batas(batas − 1)/2 = 5 × 4 / 2 = **10**. Pada baris terakhir, `range(1, 1)` kosong sehingga tidak ada titik yang dicetak.

### Latihan 8: `nested_for4_2611533018.py`

**Soal:** Buatlah program yang menerima tinggi pola berupa bilangan genap, lalu mencetak pola berbingkai yang berisi tanda `<>` yang bergerak menyempit dan melebar serta titik-titik pengisi di antaranya. Bila masukan ganjil, tampilkan pesan kesalahan.

**Kode Jawaban:**

````python
# Buat file dengan nama nested_for4_2611533018.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input ()

tinggi_3018 = int(input("Masukkan tinggi pola (bilangan genap, misal 10) : "))

if tinggi_3018 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_3018 = tinggi_3018
    c_3018 = a_3018
    lebar_3018 = (2 * tinggi_3018) - 2

    for i_3018 in range(1, tinggi_3018 + 1):
        b_3018 = c_3018 + 1

        for j_3018 in range(1, lebar_3018 + 1):

            # baris atas dan bawah
            if i_3018 == 1 or i_3018 == tinggi_3018:
                if j_3018 == 1 or j_3018 == lebar_3018:
                    print("#", end="")
                else:
                    print("=", end= "")

            # baris isi
            else:
                if j_3018 == 1 or j_3018 == lebar_3018:
                    print("|", end="")
                else:
                    if j_3018 == c_3018:
                        print("<", end="")
                    elif j_3018 == b_3018:
                        print(">", end="")
                    elif j_3018 == (lebar_3018 - c_3018):
                        print("<", end="")
                    elif j_3018 == (lebar_3018 - c_3018 + 1):
                        print(">", end= "")
                    elif j_3018 > b_3018 and j_3018 < (lebar_3018 - c_3018):
                        print(".", end="")
                    else :
                        print(" ", end="") 
                        
        print()

        # perbarui a_3018 dan c_3018 untuk baris berikutnya
        a_3018 -= 2

        if a_3018 <= 0:
            c_3018 = (-a_3018) + 2
        else:
            c_3018 = a_3018
````

**Uji Coba 2 (masukan ganjil, tinggi = 7)**

```text
Masukkan tinggi pola (bilangan genap, misal 10) : 7
Tinggi harus bilangan genap!
```

**Penjelasan:**

* **Validasi masukan.** Kondisi `tinggi_3018 % 2 != 0` mengecek bilangan ganjil dan menampilkan pesan kesalahan, sedangkan bilangan genap diproses.
* **Ukuran pola.** Pola terdiri atas `tinggi_3018` baris dan `lebar_3018 = 2 * tinggi_3018 - 2` kolom. Untuk tinggi 10, lebarnya 18 kolom.
* **Struktur nested for.** Perulangan luar `i_3018` (1 sampai tinggi) menentukan baris, perulangan dalam `j_3018` (1 sampai lebar) menentukan kolom. Total ada tinggi × lebar = 10 × 18 = **180** karakter yang diputuskan oleh rangkaian `if-elif-else`.
* **Aturan pemilihan karakter** pada setiap sel (i, j):

| Kondisi | Karakter |
|---|---|
| baris pertama atau terakhir, kolom paling kiri atau kanan | `#` |
| baris pertama atau terakhir, kolom lainnya | `=` |
| baris isi, kolom paling kiri atau kanan | `\|` |
| `j == c` atau `j == lebar − c` | `<` |
| `j == c + 1` atau `j == lebar − c + 1` | `>` |
| `c + 1 < j < lebar − c` | `.` (titik pengisi) |
| selainnya | spasi |

* **Pergerakan posisi `<>`.** Variabel `a_3018` dimulai dari `tinggi` lalu berkurang 2 di akhir setiap baris, dan variabel `c_3018` menentukan kolom tanda `<`. Selama `a_3018 > 0` berlaku `c_3018 = a_3018`, dan setelah `a_3018 <= 0` berlaku `c_3018 = -a_3018 + 2`. Nilai `c_3018` ini menyebabkan tanda `<>` pertama menyempit lalu melebar secara simetris:

| baris (i) | 2 | 3 | 4 | 5 | 6 | 7 | 8 | 9 |
|:-:|:-:|:-:|:-:|:-:|:-:|:-:|:-:|:-:|
| a sebelum dipakai | 8 | 6 | 4 | 2 | 0 | −2 | −4 | −6 |
| c (kolom `<` kiri) | 8 | 6 | 4 | 2 | 2 | 4 | 6 | 8 |
| kolom `<` kanan (lebar − c) | 10 | 12 | 14 | 16 | 16 | 14 | 12 | 10 |
| jumlah titik pengisi | 0 | 4 | 8 | 12 | 12 | 8 | 4 | 0 |

* Jumlah titik pengisi pada baris dengan nilai *c* adalah (lebar − c) − (c + 1) − 1 = **lebar − 2c − 2**. Misalnya untuk c = 2: 18 − 4 − 2 = 12 titik, sesuai output baris ke-5 dan ke-6.
* Baris 5 dan baris 6 identik karena nilai `c_3018` sama-sama 2, dan kondisi `a_3018 <= 0` menyebabkan arah pergerakan berbalik, sehingga pola simetris terhadap garis horizontal tengah.

# C.3 Latihan Segitiga

### Latihan 9: `latihan5_2611533018.py`

**Soal:** Buatlah program yang mencetak segitiga bintang rata kanan dengan tinggi sesuai masukan pengguna.

**Kode Jawaban:**

````python
# Buat file dengan nama latihan5_2611533018.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input ()

tinggi_3018 = int(input("Masukkan Tinggi Segitiga: "))

for i_3018 in range(1, tinggi_3018 + 1):

    for j_3018 in range(tinggi_3018 - i_3018):
        print("", end=" ")

    for j_3018 in range(i_3018):
        print("*", end=" ")

    print()
````

**Penjelasan:**

* Perulangan luar `i_3018` (1 sampai tinggi) menentukan nomor baris.
* Perulangan dalam pertama, `range(tinggi_3018 - i_3018)`, mencetak spasi sebagai pendorong ke kanan. Pencetakan dilakukan dengan `print("", end=" ")` sehingga setiap iterasi menghasilkan satu spasi. Jumlahnya **tinggi − i**, sehingga berkurang satu setiap baris.
* Perulangan dalam kedua, `range(i_3018)`, mencetak `*` sebanyak **i** pada baris ke-*i*, sehingga bertambah satu setiap baris.
* Pada baris ke-*i*, jumlah sel spasi + jumlah sel bintang selalu (tinggi − i) + i = **tinggi**, sehingga sisi kanan segitiga lurus. Total bintang = 1 + 2 + ... + tinggi = tinggi(tinggi + 1)/2 = 5 × 6 / 2 = **15** untuk masukan 5.
* Variabel `j_3018` dipakai ulang pada kedua perulangan dalam, hal ini aman karena kedua perulangan dijalankan berurutan dan nilainya tidak dipakai lagi setelah perulangan selesai.
